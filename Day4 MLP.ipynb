```python
import torch
from torch import nn
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms

torch.manual_seed(42)

transform = transforms.ToTensor()

train_dataset = datasets.FashionMNIST(
    root="./data",
    train=True,
    transform=transform,
    download=True
)
```

```python
indices = torch.randperm(len(train_dataset))[:64]
small_dataset = Subset(train_dataset, indices)

small_loader = DataLoader(
    small_dataset,
    batch_size=64,
    shuffle=True
)
```

```python
x, y = next(iter(small_loader))

print(x.shape)
print(y.shape)
print(y)
print(torch.bincount(y, minlength=10))
```

```python
model = nn.Sequential(
    nn.Flatten(),
    nn.Linear(28*28,256),
    nn.ReLU(),
    nn.Linear(256,10)
)
```

```python
logits = model(x)

print(logits.shape)
print(logits[0])
```

```python
predictions = logits.argmax(dim=1)
accuracy = (predictions == y).float().mean().item()
print("训练前准确率：", accuracy)
```

```python
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
for epoch in range(300):
    model.train()

    for x,y in small_loader:
        logits = model(x)
        loss = loss_fn(logits,y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    if (epoch + 1) % 20 == 0:
        model.eval()

        with torch.no_grad():
            logits = model(x)
            predictions = logits.argmax(dim=1)
            accuracy = (predictions == y).float().mean().item()

        print(
            f"epoch={epoch + 1:3d}, "
            f"loss={loss.item():.4f}, "
            f"accuracy={accuracy:.4f}"
        )
```
